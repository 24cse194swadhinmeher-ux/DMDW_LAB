```python
import numpy as np
import pandas as pd
```

```python
from sklearn.preprocessing import MinMaxScaler, StandardScaler
```

```python
data=pd.DataFrame({
    'age':[15,16,17,18,19,20],
    'studyTime':[1,2,3,2,4,3],
    'faliures':[0,1,0,2,1,0]
}).astype(float)
```

### Min-Max Normalization

```python
scaler = MinMaxScaler()
min_max_normalized_data = scaler.fit_transform(data)
min_max_normalized_df = pd.DataFrame(min_max_normalized_data, columns=data.columns)

print("Original Data:\n", data)
print("\nMin-Max Normalized Data:\n")
display(min_max_normalized_df)
```

### Z-score Normalization

```python
scaler = StandardScaler()
z_score_normalized_data = scaler.fit_transform(data)
z_score_normalized_df = pd.DataFrame(z_score_normalized_data, columns=data.columns)


print("Original Data:\n", data)
print("\nZ-score Normalized Data:\n")
display(z_score_normalized_df)
```

### Decimal Scaling Normalization

```python
def decimal_scaling_normalize(df):
    df_scaled = df.copy()
    for column in df.columns:
        max_abs_value = df[column].abs().max()
        if max_abs_value == 0:
            df_scaled[column] = 0.0
        else:

            j = np.ceil(np.log10(max_abs_value))
            df_scaled[column] = df[column] / (10 ** j)
    return df_scaled

decimal_scaled_df = decimal_scaling_normalize(data)

print("Original Data:\n", data)
print("\nDecimal Scaled Data:\n")
display(decimal_scaled_df)
```

### Summary of Normalization Results

```python
print("Original Data:\n")
display(data)

print("\nMin-Max Normalized Data:\n")
display(min_max_normalized_df)

print("\nZ-score Normalized Data:\n")
display(z_score_normalized_df)

print("\nDecimal Scaled Data:\n")
display(decimal_scaled_df)
```

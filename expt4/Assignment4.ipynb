```python
import pandas as pd
import numpy as np

data=pd.DataFrame({
    'color':[ 'red' , 'blue' , 'Red' , 'Green '],
    'City':['Delhi', 'Mumbai' , 'Delhi' , 'kolkata' ]
})
print("---Nominal Dataset -----")
print(data)

n=len(data)
matrix=np.zeros((n,n))

for i in range(n):
  for j in range(n):
    count=0

    for col in data.columns:
      if data.loc[i,col]!=data.loc[j,col]:
        count+=1
    matrix[i,j]=count/len(data.columns)
print("\n dissimilarity matrix ---  nominal")
print(pd.DataFrame(
    matrix,
    index=['01' , '02' , '03' , '04' ],
    columns=['01' , '02' , '03' , '04' ]
).round(2))
```

```python
import pandas as pd
import numpy as np

data=pd.DataFrame({
    'Student':[1,1,0,0],
    'Employee':[0,1,1,0],
    'Purchased':[1,1,0,1]
})
print("------------Binary Dataset----------")
print(data)

n=len(data)
matrix=np.zeros((n,n))

for i in range(n):
  for j in range(n):
    diff=0;

    for col in data.columns:
      if data.loc[i,col]!=data.loc[j,col]:
        diff+=1
    matrix[i,j]=diff/len(data.columns)
print("\n dissimilarity matrix ---  Binary")
print(pd.DataFrame(
    matrix,
    index=['01' , '02' , '03' , '04' ],
    columns=['01' , '02' , '03' , '04' ]
).round(2))

```

```python
import pandas as pd
import numpy as np

data=pd.DataFrame({
    'Satisfication':['low' , 'medium' , 'high' , 'medium']
})

print('----ordinal data-----')
print(data)

rank={
    'low':0,
    'medium':1,
    'high':2
}

n=len(data)
matrix=np.zeros((n,n))

for i in range(n):
  for j in range(n):

    x=rank[data.loc[i,'Satisfication']]
    y=rank[data.loc[j,'Satisfication']]
    matrix[i,j]=abs(x-y)/2
print('\n Dissimilarity matrix ---- Ordinal')
print(pd.DataFrame(
    matrix,
    index=['01' , '02' , '03' , '04' ],
    columns=['01' , '02' , '03' , '04' ]

).round(2))
```
